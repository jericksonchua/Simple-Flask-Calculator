```python
"""
Author : Precious Denisse Espaldon
Section : 9 Mendelivium
Code : Zodiac Signs
September 3, 2026
This program gives the user their zodiac sign based on the given birth year.
"""

def Zodiacyear(year):
  if year >= 1900:
  #checks if the year given by the user is valid or invalid
    zodiac = year % 12
    #divides the year by 12 to get a remainder
    match zodiac:
    #based on the remainder, it will give the user their output
    #so if the remainder is 3, the output would be case 3 : Pig
      case 0:
        print("Your Zodiac Sign is : Monkey (猴 / Hóu)")
      case 1:
        print("Your Zodiac Sign is : Rooster (鸡 / Jī)")
      case 2:
        print("Your Zodiac Sign is : Dog (狗 / Gǒu)")
      case 3:
        print("Your Zodiac Sign is : Pig (猪 / Zhū)")
      case 4:
        print("Your Zodiac Sign is : Rat (鼠 / Shǔ)")
      case 5:
        print("Your Zodiac Sign is : Ox (牛 / Niú)")
      case 6:
        print("Your Zodiac Sign is : Tiger (虎 / Hǔ)")
      case 7:
        print("Your Zodiac Sign is : Rabbit (兔 / Tù)")
      case 8:
        print("Your Zodiac Sign is : Dragon (龙 / Lóng)")
      case 9:
        print("Your Zodiac Sign is : Snake (蛇 / Shé)")
      case 10:
        print("Your Zodiac Sign is : Horse (马 / Mǎ)")
      case 11:
        print("Your Zodiac Sign is : Goat (羊 / Yáng)")
  else:
    print("Invalid input, year should not be earlier than 1900")

birth_year = int(input("Enter your birth year : "))
Zodiacyear(birth_year)
```
